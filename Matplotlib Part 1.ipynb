```python
batsman=pd.read_csv("~/Machine Learning/Matplotlib/Datasets/sharma-kohli.csv")
batsman
```

```python
plt.plot(batsman['index'],batsman['RG Sharma'])
```

```python
#plotting multiple graphs in a single plot
plt.plot(batsman['index'],batsman['RG Sharma'])
plt.plot(batsman['index'],batsman['V Kohli'])

#labelling x and y axis
plt.xlabel("Season")
plt.ylabel("Runs")

#Title
plt.title("Virat Kohli vs Rohit Sharma Comparison Season by Season")
```

#### Colors and styling

```python
#linestyle and colour
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle=':')
plt.plot(batsman['index'],batsman['V Kohli'],color='#0000FF')
```

```python
#Linewidth
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',linestyle='-.')
plt.plot(batsman['index'],batsman['V Kohli'],color='#0000FF',linewidth=5)
```

```python
#marker and markersize
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',marker='*')
plt.plot(batsman['index'],batsman['V Kohli'],color='#0000FF',marker='o',markersize=15)
```

#### Labelling and legend() function

```python
plt.plot(batsman['index'],batsman['RG Sharma'],color='red',marker='o',label='Rohit Sharma')
plt.plot(batsman['index'],batsman['V Kohli'],color='#0000FF',marker='o',label='Virat Kohli')

#labels for the lines wont show unless we call the legend() function
plt.legend(loc='center left') #by default loc value is 'best' (it will automatically figure out the best place)
```

#### Limiting axes for handling outliers

```python
# an outlier will make our graph look flat for other values 
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]

plt.plot(year,price)
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]
# ylim and xlim will limit the y and x axis values respectively
plt.plot(year,price)
plt.ylim(0,100000)
```

```python
# limiting axes
price = [48000,54000,57000,49000,47000,45000,4500000]
year = [2015,2016,2017,2018,2019,2020,2021]
# ylim and xlim will limit the y and x axis values respectively
plt.plot(year,price)
plt.ylim(0,100000)
plt.xlim(2015,2019)
```

#### grid()

```python
## grid function will turn on the grid

plt.plot(batsman['index'],batsman['RG Sharma'],label='R Sharma',marker='o')
plt.plot(batsman['index'],batsman['V Kohli'],label='V Kohli',marker='o')
plt.legend()
plt.grid()

plt.xlabel("Season")
plt.ylabel("Runs")


plt.title("Virat Kohli vs Rohit Sharma Comparison Season by Season")
```

```python
plt.figure(figsize=(10,6)) # figsize will be 10 on x axis and 6 on y axis
internet_speed=pd.read_csv("~/Machine Learning/Matplotlib/Datasets/results.csv")
internet_speed=internet_speed[internet_speed['ConnType']=='Wifi']
internet_speed['Date']=internet_speed['Date'].str.split(',').str.get(1)
plt.plot(internet_speed['Date'],internet_speed['Download Speed'],color='red',marker='o')
plt.gca().invert_xaxis() ## this will invert x axis
```

#### show()

```python
# show function will plot the graphs actually but jupyter notebook shows the graphs without this function
plt.show()
```

```python
df=pd.read_csv("~/Machine Learning/Matplotlib/Datasets/batter.csv")
df
```

```python
top_50=df.head(50)
plt.scatter(top_50['avg'],top_50['strike_rate'])
plt.xlabel("Average")
plt.ylabel("Strike Rate")
plt.title("Strike Rate vs Average Analysis")
```

```python
# We can also pass the 's' parameter to change the size of the dot according to the value of the strike rate
plt.scatter(top_50['avg'],top_50['strike_rate'],s=5*(top_50['strike_rate']-100),color='red',marker='o')
plt.xlabel("Average")
plt.ylabel("Strike Rate")
plt.title("Strike Rate vs Average Analysis")
```

### Stacked Bar Chart

```python
plt.bar(df['batsman'],df['2015'],label='2015')
plt.bar(df['batsman'],df['2016'],label='2016',bottom=df['2015'])
plt.bar(df['batsman'],df['2017'],label='2017',bottom=(df['2015']+df['2016']))
plt.legend()
```
